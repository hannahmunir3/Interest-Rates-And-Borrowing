## Reading data

```python
df = pd.read_csv(r"D:\Github\Interest-Rates-And-Borrowing\Data\household_finances.csv")
```

```python
df["Date"] = df["Date"].astype(str)
```

# Bank rate line graph

```python
plt.figure(figsize=(10,5))

# Plotting data
plt.plot(df["Date"], df["Bank rate: %"], marker="o", color="red")

# Creating titles/ labels
plt.title("UK Bank Rate, 2019-2025")
plt.xlabel("Quarterly data, 2019-2025")
plt.ylabel("Bank Rate (%)")
#
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

# Debt to Income ratio line graph

```python
plt.figure(figsize=(10,5))

# Plotting data
plt.plot(df["Date"], df["Debt to Income ratio: %:"], marker="o", color="darkblue")

# Creating titles/ labels
plt.title("UK Debt to Income Ratio, 2019-2025")
plt.xlabel("Quarterly data, 2019-2025")
plt.ylabel("Debt to Income ratio: %")
#
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

# Savings Ratio line graph

```python
plt.figure(figsize=(10,5))

# Plotting data
plt.plot(df["Date"], df["Households savings ratio per head"], marker="o", color="darkgreen")

# Creating titles/ labels
plt.title("UK Household Savings Ratio, 2019-2025")
plt.xlabel("Quarterly data, 2019-2025")
plt.ylabel("Savings Ratio (%)")
#
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
